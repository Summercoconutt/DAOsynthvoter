```python
# load a parquet file into a pandas dataframe
import pandas as pd

df = pd.read_parquet('/home/carlo/codes/DAOsynthvoter/data/processed/votes_cleaned_from_server.parquet')
```

```python
df['Proposal Body'].unique()
```

```python
#from column Proposal Body, select the unique values that are at least 100 characters long and store them in a list
unique_proposals = df['Proposal Body'].unique()
long_proposals = [proposal for proposal in unique_proposals if len(proposal) >= 100]
```

```python
#calculate the summary statistics of the length of the proposals in long_proposals
import numpy as np
summary_stats = np.ptp(np.array([len(proposal) for proposal in long_proposals]))
```

```python
prop_charlength = np.array([len(proposal) for proposal in long_proposals])
```

```python
# calcualte the mean, median, stdev, IQR, min and max of the length of the proposals in long_proposals
mean_length = np.mean(prop_charlength)
median_length = np.median(prop_charlength)
stdev_length = np.std(prop_charlength)
iqr_length = np.percentile(prop_charlength, 75) - np.percentile(prop_charlength, 25)
min_length = np.min(prop_charlength)
max_length = np.max(prop_charlength)

#print the summary statistics

print(f'Mean length: {mean_length}')
print(f'Median length: {median_length}')
print(f'Standard deviation: {stdev_length}')
print(f'Interquartile range: {iqr_length}')
print(f'Minimum length: {min_length}')
print(f'Maximum length: {max_length}')
```
